**Tasks:**

```python
from standards import *
import copy
C_POS = Vector3(5504501.979, 4407131.291, 614552.796)
C_VEL = Vector3(-4328.607517, 4891.156799, 3695.139757)
D_POS = Vector3(5495520.274, 4406811.959, 616985.925)
D_VEL = Vector3(-4336.950279, 4892.708100, 3698.323755)
C_KEP = KeplerianElements(C_POS, C_VEL)
D_KEP = KeplerianElements(D_POS, D_VEL)
Earth_gravitational_parameter = 3.986004418e14 #m^3/s^2
sma_diff = D_KEP.a - C_KEP.a
ecc_diff = D_KEP.ecc - C_KEP.ecc
inc_diff = D_KEP.inc_deg - C_KEP.inc_deg
raan_dif = D_KEP.raan_deg - C_KEP.raan_deg
argp_diff = D_KEP.argp_deg - C_KEP.argp_deg
ta_diff = D_KEP.ta_deg - C_KEP.ta_deg
print(f"SMA Diff:  {sma_diff:.6f} m")
print(f"ECC Diff:  {ecc_diff:.6f}")
print(f"INC Diff:  {inc_diff:.6f} deg")
print(f"RAAN Diff: {raan_dif:.6f} deg")
print(f"ARGP Diff: {argp_diff:.6f} deg")
print(f"TA Diff:   {ta_diff:.6f} deg")
C_ECI_TO_UVW = compute_matrix_eci_to_uvw(C_POS, C_VEL)
diff_ECI = D_POS - C_POS
d_UVW_POS = C_ECI_TO_UVW @ diff_ECI.get_np_vector()
print("UVW diff: \n" + str(d_UVW_POS))
C_MM = sqrt(C_KEP.mu_earth/(C_KEP.a**3)) #rad/sec
print(f"Chief Mean Motion: {C_MM:.6f} rad/sec, or {(C_MM)/(2*pi)*(60*60*24):.3f} rotations/day")
n = C_MM
t = 1
HCW_MATRIX = np.array([
    [4 - 3*np.cos(n*t),0,0,(1/n)*np.sin(n*t),(2/n)*(1 - np.cos(n*t)),0],
    [6*(np.sin(n*t) - n*t),1,0,(-2/n)*(1 - np.cos(n*t)),(1/n)*(4*np.sin(n*t) - 3*n*t),0],
    [0,0,np.cos(n*t),0,0,(1/n)*np.sin(n*t)],
    [3*n*np.sin(n*t),0,0,np.cos(n*t),2*np.sin(n*t),0],
    [-6*n*(1 - np.cos(n*t)),0,0,-2*np.sin(n*t),4*np.cos(n*t) - 3,0],
    [0,0,-n*np.sin(n*t),0,0,np.cos(n*t)]
])
print("HCW Matrix for t=1:\n" + str(HCW_MATRIX))

#propagate 6000 sec at 60 sec step size
class Step:
    def __init__(self, T, V1_RK_X, V1_RK_Y, V1_RK_Z, V1_RK_XD, V1_RK_YD, V1_RK_ZD,
                 V2_RK_X, V2_RK_Y, V2_RK_Z, V2_RK_XD, V2_RK_YD, V2_RK_ZD, V1_RK_U,
                 V1_RK_V, V1_RK_W, V2_RK_U, V2_RK_V, V2_RK_W):
        self.T = T
        self.V1_RK_X = V1_RK_X
        self.V1_RK_Y = V1_RK_Y
        self.V1_RK_Z = V1_RK_Z
        self.V1_RK_XD = V1_RK_XD
        self.V1_RK_YD = V1_RK_YD
        self.V1_RK_ZD = V1_RK_ZD
        self.V2_RK_X = V2_RK_X
        self.V2_RK_Y = V2_RK_Y
        self.V2_RK_Z = V2_RK_Z
        self.V2_RK_XD = V2_RK_XD
        self.V2_RK_YD = V2_RK_YD
        self.V2_RK_ZD = V2_RK_ZD
        self.V1_RK_U = V1_RK_U
        self.V1_RK_V = V1_RK_V
        self.V1_RK_W = V1_RK_W
        self.V2_RK_U = V2_RK_U
        self.V2_RK_V = V2_RK_V
        self.V2_RK_W = V2_RK_W

steps_central_body:list[Step] = []
V1_y_0 = Vector6(C_POS.x, C_POS.y, C_POS.z, C_VEL.x, C_VEL.y, C_VEL.z)
V2_y_0 = Vector6(D_POS.x, D_POS.y, D_POS.z, D_VEL.x, D_VEL.y, D_VEL.z)
h = 60
step_num = 100
w_ = Vector3(0, 0, 72.921151467e-6)
V1_v_r_ = C_VEL - (w_.cross(C_POS))
V2_v_r_ = D_VEL - (w_.cross(D_POS))
V1_temp_pos = copy.deepcopy(C_POS)
V2_temp_pos = copy.deepcopy(D_POS)
V1_temp_vel = copy.deepcopy(C_VEL)
V2_temp_vel = copy.deepcopy(D_VEL)
start_time = datetime(year=2024, month=4, day=9,hour=0, minute=0, second=0)
timestamp = start_time

matrix = compute_matrix_eci_to_uvw(V1_temp_pos, V1_temp_vel)
V1_UVW_np = matrix @ V1_temp_pos.get_np_vector()
V2_UVW_np = matrix @ V2_temp_pos.get_np_vector()

steps_central_body.append(Step(0, V1_y_0.x, V1_y_0.y, V1_y_0.z, V1_temp_vel.x, V1_temp_vel.y, V1_temp_vel.z,
V2_y_0.x, V2_y_0.y, V2_y_0.z, V2_temp_vel.x, V2_temp_vel.y, V2_temp_vel.z, V1_UVW_np[0][0], V1_UVW_np[1][0], V1_UVW_np[2][0], V2_UVW_np[0][0], V2_UVW_np[1][0], V2_UVW_np[2][0]))

for i in range(step_num):
    # calculate TOD position vectors
    t = h*(i+1)
    timestamp = timestamp + timedelta(seconds=h)
    V1_step = compute_rk_2_body_step(0, Earth_gravitational_parameter, V1_temp_pos, V1_temp_vel, h, V1_y_0, timestamp, 0, 0, 0)
    V2_step = compute_rk_2_body_step(0, Earth_gravitational_parameter, V2_temp_pos, V2_temp_vel, h, V2_y_0, timestamp, 0, 0, 0)
    matrix = compute_matrix_eci_to_uvw(V1_temp_pos, V1_temp_vel)
    V1_UVW_np = matrix @ V1_temp_pos.get_np_vector()
    V2_UVW_np = matrix @ V2_temp_pos.get_np_vector()
    # append data to array and move on
    steps_central_body.append(Step(t, V1_step.x, V1_step.y, V1_step.z, V1_step.xd, V1_step.yd, V1_step.zd,
        V2_step.x, V2_step.y, V2_step.z, V2_step.xd, V2_step.yd, V2_step.zd, V1_UVW_np[0][0], V1_UVW_np[1][0], V1_UVW_np[2][0], V2_UVW_np[0][0], V2_UVW_np[1][0], V2_UVW_np[2][0]))
    V1_y_0 = V1_step
    V2_y_0 = V2_step
    V1_temp_pos = Vector3(V1_step.x, V1_step.y, V1_step.z)
    V1_temp_vel = Vector3(V1_step.xd, V1_step.yd, V1_step.zd)
    V2_temp_pos = Vector3(V2_step.x, V2_step.y, V2_step.z)
    V2_temp_vel = Vector3(V2_step.xd, V2_step.yd, V2_step.zd)


from IPython.display import Markdown, display
header = "| Step | Time (s) | V1_X (m) | V1_Y (m) | V1_Z (m) | V1_XD (m/s) | V1_YD (m/s) | V1_ZD (m/s) | V2_X (m) | V2_Y (m) | V2_Z (m) | V2_XD (m/s) | V2_YD (m/s) | V2_ZD (m/s) | Diff_U (m) | Diff_V (m) | Diff_W (m) |"
separator = "|------|----------|----------|----------|----------|--------------|--------------|--------------|----------|----------|----------|--------------|--------------|--------------|----------|----------|----------|"

rows = []
for i, s in enumerate(steps_central_body):
    row = (
        f"| {i} | {s.T:.0f} | "
        f"{s.V1_RK_X:.2f} | {s.V1_RK_Y:.2f} | {s.V1_RK_Z:.2f} | "
        f"{s.V1_RK_XD:.2f} | {s.V1_RK_YD:.2f} | {s.V1_RK_ZD:.2f} | "
        f"{s.V2_RK_X:.2f} | {s.V2_RK_Y:.2f} | {s.V2_RK_Z:.2f} | "
        f"{s.V2_RK_XD:.2f} | {s.V2_RK_YD:.2f} | {s.V2_RK_ZD:.2f} | "
        f"{(s.V2_RK_U-s.V1_RK_U):.2f} | {(s.V2_RK_V-s.V1_RK_V):.2f} | {(s.V2_RK_W-s.V1_RK_W):.2f} | "
    )
    rows.append(row)

table_md = "\n".join([header, separator] + rows)
display(Markdown(f"### Propagation Table\n\n{table_md}"))

import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator

diff_u = []
diff_v = []

for s in steps_central_body:
    du = s.V2_RK_U - s.V1_RK_U
    dv = s.V2_RK_V - s.V1_RK_V
    diff_u.append(du)
    diff_v.append(dv)

plt.figure(figsize=(10, 8))

# Main plot: ΔU (radial) vs ΔV (in-track)
plt.plot(diff_v, diff_u, 'b-', linewidth=2, label='Chief Vs Deputy')

# Flip x-axis so negative values are on the RIGHT
plt.gca().invert_xaxis()

plt.xlabel('Intrack Diff [m]', fontsize=14)
plt.ylabel('Radial Diff [m]', fontsize=14)
plt.title('Relative Motion: Radial vs Intrack', fontsize=16)
plt.axhline(0, color='black', linewidth=0.8, linestyle='-')
plt.axvline(0, color='black', linewidth=0.8, linestyle='-')
plt.gca().yaxis.set_major_locator(MultipleLocator(2000))

plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=12)
plt.axis('equal')          # keeps the relative motion shape correct
plt.tight_layout()

plt.show()
```

**Using the .CSV file provided** (`Homework 13 Problem 2.csv` — see Slide 7 for reading instructions):

- Use the **ECEF** values to estimate the time of the **first two ascending node crossings**  
  ($z_{\text{ECEF}}$ transitions from negative to positive).

- Using $x_{\text{ECEF}}$ and $y_{\text{ECEF}}$ at the zero-crossing time, compute the  
  **geographic longitude of ascending node (GLAN)** for each crossing.

- Estimate the **orbit period** (difference between node crossing times).

- Estimate the **GLAN rate** (difference between GLAN values divided by period).

**Tips:**

- You can do this by eye in Excel using linear interpolation.
- You can also do it in MATLAB:
  - Read the file as shown on Slide 7.
  - Search for $z_{\text{ECEF}}$ negative-to-positive crossings.

```python
from dataclasses import dataclass
import csv
from typing import List

@dataclass
class EphemerisPoint:
    time: datetime
    j2000_x: float
    j2000_y: float
    j2000_z: float
    j2000_xd: float
    j2000_yd: float
    j2000_zd: float
    j2000_ax: float
    j2000_ay: float
    j2000_az: float
    x_ecef: float
    y_ecef: float
    z_ecef: float
    xd_ecef: float
    yd_ecef: float
    zd_ecef: float

def load_ephemeris(filename: str = "Homework_13_Problem_2.csv") -> List[EphemerisPoint]:
    data: List[EphemerisPoint] = []
    
    with open(filename, newline='') as f:
        reader = csv.reader(f)
        next(reader)  # skip the header row
        
        for row in reader:
            if not row:  
                continue
                
            data.append(EphemerisPoint(
                time=datetime.strptime(row[0], "%m/%d/%y %H:%M:%S"),
                j2000_x=float(row[1]),
                j2000_y=float(row[2]),
                j2000_z=float(row[3]),
                j2000_xd=float(row[4]),
                j2000_yd=float(row[5]),
                j2000_zd=float(row[6]),
                j2000_ax=float(row[7]),
                j2000_ay=float(row[8]),
                j2000_az=float(row[9]),
                x_ecef=float(row[10]),
                y_ecef=float(row[11]),
                z_ecef=float(row[12]),
                xd_ecef=float(row[13]),
                yd_ecef=float(row[14]),
                zd_ecef=float(row[15])
            ))
    return data
points = load_ephemeris()

# nodal crossings:
neg_to_pos_crossings = []
state = 'BELOW'
if points[0].z_ecef > 0:
    state='ABOVE'
for i in range(len(points)):
    if points[i].z_ecef >= 0 and state=='BELOW':
        neg_to_pos_crossings.append([points[i-1], points[i]])
    if points[i].z_ecef > 0:
        state = 'ABOVE'
    else:
        state = 'BELOW'
print(str(len(neg_to_pos_crossings)) + " nodal crossings")
p1 = neg_to_pos_crossings[0][0]
p2 = neg_to_pos_crossings[0][1]
expected_z_value = 0
fraction = (expected_z_value - p1.z_ecef) / (p2.z_ecef - p1.z_ecef)
interp_time = p1.time + fraction * (p2.time - p1.time)
print("Crossing 1: " + str(interp_time))
p1 = neg_to_pos_crossings[1][0]
p2 = neg_to_pos_crossings[1][1]
expected_z_value = 0
fraction = (expected_z_value - p1.z_ecef) / (p2.z_ecef - p1.z_ecef)
interp_time2 = p1.time + fraction * (p2.time - p1.time)
print("Crossing 2: " + str(interp_time2))
times = [p.time for p in points]
z_vals = [p.z_ecef for p in points]
plt.figure(figsize=(12, 4))  # (width, height) in inches
plt.plot(times, z_vals)
plt.xlabel("Time")
plt.ylabel("z_ECEF (m)")
plt.scatter(interp_time, 0, color='red', zorder=3)
plt.scatter(interp_time2, 0, color='red', zorder=3)
plt.title("z_ECEF vs Time")
plt.grid()
plt.show()

#longitude of asc node (GLAN)
print(points[0].)
```
