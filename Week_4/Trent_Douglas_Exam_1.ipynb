```python
from standards import *
import math
pos_ = Vector3(-17423040.106245, -12809407.333084, -1928704.825456)
vel_ = Vector3(-66.389180, -2939.847170, -2034.815392)

kep_elements = KeplerianElements(pos_, vel_)
print(f"{'Semi-major axis (a)':<30} {kep_elements.a:16.8f} m")
print(f"{'Eccentricity (e)':<30} {kep_elements.ecc:16.8f}")
print(f"{'Inclination (i)':<30} {math.degrees(kep_elements.inc):16.8f} deg")
print(f"{'RAAN (Ω)':<30} {math.degrees(kep_elements.raan):16.8f} deg")
print(f"{'Argument of periapsis (ωp)':<30} {math.degrees(kep_elements.argp):16.8f} deg")
print(f"{'True anomaly (ν)':<30} {math.degrees(kep_elements.ta):16.8f} deg")
```

For steps **c.** and **d.**, please present the combined results in tabular form with heading as follows:

| v          | v + Δv     | f     | g     | ḟ̇    | ġ̇    | x(Δv)     | y(Δv)     | ẋ(Δv)     | ẏ(Δv)     | Δt    |
|------------|------------|-------|-------|-------|-------|-----------|-----------|-----------|-----------|-------|

where the x(Δv), etc., terms correspond to the components of the perifocal vectors on the **Left-Hand side** of the equations:

$\vec{r}_p(\Delta v) = \begin{bmatrix} x(\Delta v) \\ y(\Delta v) \\ 0 \end{bmatrix} = f \vec{r}_{0,p} + g \dot{\vec{r}}_{0,p}$

$\dot{\vec{r}}_p(\Delta v) = \begin{bmatrix} \dot{x}(\Delta v) \\ \dot{y}(\Delta v) \\ 0 \end{bmatrix} = \dot{f} \vec{r}_{0,p} + \dot{g} \dot{\vec{r}}_{0,p}$

#### **c.** Beginning at perigee (v₀ = 0), compute the f and g functions and perifocal position and velocity for 3 consecutive steps using Δv = 72° (24 points)

- Note that the vector provided is **not** at perigee. You will first need to find the position and velocity vectors **at perigee**. See, e.g., Slide 39 of the Class 2 notes, and use the value of true anomaly ν at perigee.
- You can choose from a couple of approaches:
  - Use the end vectors for each step, as the beginning vectors for the next step of Δv = 72°.
  - Use the cumulative angle from perigee and compute f and g functions from perigee.
  - Make sure your angles are in **radians**.

For each of the 3 steps, compute:

- the Lagrange coefficients **f** and **g**
- the updated **perifocal position vector** r̅_p
- the updated **perifocal velocity vector** v̅_p

#### **d.** For each step, compute the time of flight (in seconds) to travel the angle Δv = 72° between the successive end points (12 points)
