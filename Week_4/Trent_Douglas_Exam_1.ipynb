In this problem, the input vector is assumed to be in an **inertial reference frame** where the axes are fixed at **midnight** of the day of epoch (i.e., 07 Feb 2023 00:00:00). As such, you may **approximate** the ECI → ECEF transformation as a simple rotation about the Earth’s Z-axis. The transformation matrix is given by:

$
\left[ \mathbf{T}^{\text{ECEF}}_{\text{ECI}} \right] = R_z(\omega_\oplus \Delta t)
$

Where:

- $\omega_\oplus = 7.292116 \times 10^{-5}$ rad/s  
- $\Delta t =$ seconds from **07 Feb 2023 00:00:00**

#### **a.** (5 points)  
Compute the **ECI → ECEF transformation matrix** $R_z$ corresponding to Earth’s rotation from **midnight** to the epoch of the orbit state vector.

```python
from datetime import datetime
t_0 = datetime(year=2023, month=2, day=7, hour=0, minute=0)
t_epoch = datetime(year=2026, month=2, day=7, hour=12, minute=0)
delta = t_epoch - t_0
time_delta_secs = delta.seconds # seconds
total_rotation_rad = 7.292116*math.pow(10, -5) * time_delta_secs
#rotation about z axis so we use this:
R = np.array([
    [math.cos(total_rotation_rad), math.sin(total_rotation_rad), 0],
    [-math.sin(total_rotation_rad), math.cos(total_rotation_rad), 0],
    [0, 0, 1]
])
print("Rotation matrix from ECI to ECEF:")
print(str(R))
```

#### **b.** (5 points)  
Compute the **ECEF position vector** by applying the transformation matrix from part (a) to the ECI position vector given above.


```python
ECI_pos = Vector3(-5215459.436366, -1418270.765462, 4678940.232127)
ECEF_pos_np = R@ECI_pos.get_np_vector()
ECEF_pos = Vector3(float(ECEF_pos_np[0][0]), float(ECEF_pos_np[1][0]), float(ECEF_pos_np[2][0]))
print("ECEF position vector:")
print(ECEF_pos)
```

#### **c.** (10 points)  
Compute the **longitude** $\lambda$ and **geocentric latitude** $\phi_c$ of the ECEF vector from part (b).

**Hint:**  
In this problem, $\lambda$ and $\phi_c$ are mathematically identical to **right ascension** ($\alpha$) and **declination** ($\delta$), measured with respect to the **ECEF axes**. See Lecture 3 notes for the definitions of right ascension and declination.

The right ascension $\alpha$ and declination $\delta$ are computed as:

$
\begin{aligned}
\alpha &= \operatorname{atan2}(v_j, v_i) \\
\delta &= \sin^{-1}\left(\frac{v_k}{|\vec{v}|} \right)
\end{aligned}
$

```python
ra = math.atan2(ECEF_pos.y, ECEF_pos.x)
dec = math.asin(ECEF_pos.z/abs(ECEF_pos.magnitude()))
print("Longitude: " + str(math.degrees(ra)) + " deg")
print("Latitude:  " + str(math.degrees(dec)) + " deg")
```
